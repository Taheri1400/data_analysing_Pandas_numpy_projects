```python
# یک نسخه برای آماده‌سازی داده‌ها
df_clean = df.drop(
    columns=['search_count_group'],
    errors='ignore'
).copy()

# تبدیل تاریخ ورود و خروج به datetime
df_clean['checkIn_date'] = pd.to_datetime(
    df_clean['checkIn_date'],
    errors='coerce'
)

df_clean['checkOut_date'] = pd.to_datetime(
    df_clean['checkOut_date'],
    errors='coerce'
)

# درصد مقادیر گمشده
print("Missing percentage:")
print(df_clean.isnull().mean() * 100)

# بررسی مقادیر گمشده بر اساس رزرو
print("\nMissing values by booking status:")

for col in ['destination_distance', 'checkIn_date', 'checkOut_date']:
    print(f"\n{col}")
    print(
        df_clean.groupby('is_booking')[col]
        .apply(lambda x: x.isnull().mean() * 100)
    )
```

```python
print("Check-in examples:")
print(df_clean['checkIn_date'].head(10))

print("\nCheck-out examples:")
print(df_clean['checkOut_date'].head(10))

print("\nInvalid date order:")

print(
    (
        df_clean['checkOut_date'] < df_clean['checkIn_date']
    ).sum()
)
```

```python
# فاصله زمانی تا ورود
df_clean['days_until_checkin'] = (
    df_clean['checkIn_date'] - df_clean['search_date']
).dt.total_seconds() / (24 * 3600)

# مدت اقامت
df_clean['stay_duration'] = (
    df_clean['checkOut_date'] - df_clean['checkIn_date']
).dt.days

# بررسی آماری
print("Days until check-in:")
print(df_clean['days_until_checkin'].describe())

print("\nStay duration:")
print(df_clean['stay_duration'].describe())

# بررسی مقادیر غیرعادی
print("\nNegative days until check-in:",
      (df_clean['days_until_checkin'] < 0).sum())

print("Non-positive stay duration:",
      (df_clean['stay_duration'] <= 0).sum())
```

```python
# بررسی فاصله منفی تا ورود بر اساس وضعیت رزرو
print("Negative days until check-in by booking status:")
print(
    df_clean.groupby('is_booking')['days_until_checkin']
    .apply(lambda x: (x < 0).sum())
)

# درصد فاصله‌های منفی در هر گروه
print("\nPercentage of negative values:")
print(
    df_clean.groupby('is_booking')['days_until_checkin']
    .apply(lambda x: (x < 0).mean() * 100)
)

# نمونه ردیف‌هایی با تاریخ ورود قبل از جست‌وجو
print("\nExamples of negative days:")
print(
    df_clean.loc[
        df_clean['days_until_checkin'] < 0,
        ['search_date', 'checkIn_date', 'checkOut_date',
         'days_until_checkin', 'is_booking']
    ].head(15)
)

# بررسی اقامت صفرروزه
print("\nZero-day stays by booking status:")
print(
    df_clean.groupby('is_booking')['stay_duration']
    .apply(lambda x: (x == 0).sum())
)

# نمونه اقامت‌های طولانی‌تر از 30 روز
print("\nLong stays:")
print(
    df_clean.loc[
        df_clean['stay_duration'] > 30,
        ['checkIn_date', 'checkOut_date',
         'stay_duration', 'is_booking']
    ].head(15)
)
```

```python
# حذف اثر ساعت از تاریخ جست‌وجو
df_clean['days_until_checkin'] = (
    df_clean['checkIn_date'] -
    df_clean['search_date'].dt.normalize()
).dt.days

print("Days until check-in:")
print(df_clean['days_until_checkin'].describe())

print("\nNegative values by booking status:")
print(
    df_clean.groupby('is_booking')['days_until_checkin']
    .apply(lambda x: (x < 0).sum())
)

print("\nSame-day check-in by booking status:")
print(
    df_clean.groupby('is_booking')['days_until_checkin']
    .apply(lambda x: (x == 0).sum())
)

print("\nStays longer than 30 days by booking status:")
print(
    df_clean.groupby('is_booking')['stay_duration']
    .apply(lambda x: (x > 30).sum())
)
```

```python
# نمونه تاریخ‌های واقعاً منفی
negative_dates = df_clean[
    df_clean['days_until_checkin'] < 0
]

print("Negative cases:", len(negative_dates))

print("\nExamples:")
print(
    negative_dates[
        ['search_date', 'checkIn_date', 'checkOut_date',
         'days_until_checkin', 'stay_duration', 'is_booking']
    ].head(20)
)

print("\nDistribution of negative days:")
print(negative_dates['days_until_checkin'].value_counts().sort_index())

# بررسی اقامت‌های طولانی‌تر از 30 روز
long_stays = df_clean[df_clean['stay_duration'] > 30]

print("\nLong stays by booking status:")
print(long_stays['is_booking'].value_counts())
```

```python
# حذف ستون کمکی قبلی
df_analysis = df_clean.drop(
    columns=['search_count_group'],
    errors='ignore'
).copy()

# مرتب‌سازی بر اساس کاربر و زمان
df_analysis = df_analysis.sort_values(
    ['user', 'search_date']
)

# تعریف مشخصات مشترک جست‌وجو
search_columns = [
    'user',
    'destination',
    'checkIn_date',
    'checkOut_date',
    'n_adults',
    'n_children',
    'n_rooms',
    'is_package'
]

# تعداد ردیف‌ها برای هر ترکیب مشخصات
group_sizes = df_analysis.groupby(
    search_columns,
    dropna=False
).size()

print("Number of rows per search-like group:")
print(group_sizes.describe())

print("\nGroups with more than one row:")
print((group_sizes > 1).sum())

print("\nPercentage of rows in repeated groups:")
print(
    df_analysis.groupby(
        search_columns,
        dropna=False
    )['user'].transform('size').gt(1).mean() * 100
)
```

```python
# گروه‌بندی بر اساس مشخصات جست‌وجو
search_columns = [
    'user',
    'destination',
    'checkIn_date',
    'checkOut_date',
    'n_adults',
    'n_children',
    'n_rooms',
    'is_package'
]

# محاسبه اولین و آخرین زمان هر گروه
group_times = (
    df_analysis.groupby(search_columns, dropna=False)
    .agg(
        first_time=('search_date', 'min'),
        last_time=('search_date', 'max'),
        number_of_rows=('search_date', 'size')
    )
)

# مدت زمان بین اولین و آخرین ردیف هر گروه، بر حسب ثانیه
group_times['time_span_seconds'] = (
    group_times['last_time'] - group_times['first_time']
).dt.total_seconds()

# فقط گروه‌های تکراری
repeated_groups = group_times[
    group_times['number_of_rows'] > 1
]

print("Time span of repeated groups (seconds):")
print(repeated_groups['time_span_seconds'].describe())

print("\nTime span distribution:")
print(
    repeated_groups['time_span_seconds']
    .quantile([0, .25, .5, .75, .90, .95, .99, 1])
)

print("\nGroups completed within:")
for seconds in [10, 30, 60, 300, 600, 1800]:
    percentage = (
        repeated_groups['time_span_seconds'].le(seconds).mean() * 100
    )
    print(f"{seconds} seconds: {percentage:.2f}%")
```

```python
# تعداد ردیف‌های هر گروه مشخصات جست‌وجو
df_analysis['group_size'] = (
    df_analysis.groupby(
        search_columns,
        dropna=False
    )['user'].transform('size')
)

# بررسی ارتباط search_count و group_size
print("Correlation:")
print(
    df_analysis[['search_count', 'group_size']]
    .corr()
)

print("\nGroup size statistics by search_count:")
print(
    df_analysis.groupby('search_count')['group_size']
    .agg(['count', 'mean', 'median', 'max'])
    .head(20)
)

# بررسی چند نمونه با group_size بزرگ
print("\nExamples of large groups:")
print(
    df_analysis.loc[
        df_analysis['group_size'] >= 10,
        search_columns + [
            'search_date',
            'search_count',
            'hotel_category',
            'is_booking',
            'group_size'
        ]
    ].head(20).to_string(index=False)
)
```

```python
# تعداد hotel_categoryهای متفاوت در هر گروه
category_stats = (
    df_analysis.groupby(search_columns, dropna=False)
    .agg(
        number_of_rows=('hotel_category', 'size'),
        unique_categories=('hotel_category', 'nunique'),
        number_of_bookings=('is_booking', 'sum')
    )
)

print("Unique hotel categories per group:")
print(category_stats['unique_categories'].describe())

print("\nGroups with multiple categories:")
print((category_stats['unique_categories'] > 1).sum())

print("\nGroups with at least one booking:")
print((category_stats['number_of_bookings'] > 0).sum())

print("\nGroups with more than one booking:")
print((category_stats['number_of_bookings'] > 1).sum())

print("\nExamples with multiple categories:")
print(
    category_stats[
        category_stats['unique_categories'] > 1
    ].head(15)
)
```

```python
train_end = '2014-06-30'
valid_start = '2014-07-01'
valid_end = '2014-09-30'
test_start = '2014-10-01'

train = df_clean[df_clean['search_date'] <= train_end]

valid = df_clean[
    (df_clean['search_date'] >= valid_start) &
    (df_clean['search_date'] <= valid_end)
]

test = df_clean[df_clean['search_date'] >= test_start]

for name, data in [
    ('Train', train),
    ('Validation', valid),
    ('Test', test)
]:
    print(f'\n{name}')
    print('Records:', len(data))
    print('Bookings:', data['is_booking'].sum())
    print('Booking rate:', data['is_booking'].mean())
    print('Unique users:', data['user'].nunique())
```

```python
train_users = set(train['user'].unique())
valid_users = set(valid['user'].unique())
test_users = set(test['user'].unique())

print("Train users:", len(train_users))
print("Validation users:", len(valid_users))
print("Test users:", len(test_users))

print("\nValidation:")
print("Users seen in Train:",
      len(valid_users & train_users))
print("New users:",
      len(valid_users - train_users))

print("\nTest:")
print("Users seen in Train:",
      len(test_users & train_users))
print("Users seen in Validation:",
      len(test_users & valid_users))
print("Users not seen in Train or Validation:",
      len(test_users - (train_users | valid_users)))

print("\nUsers appearing in all three sets:",
      len(train_users & valid_users & test_users))
```
