```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# تولید داده‌های نرمال (نمرات کلاس) و اضافه کردن دو نمره اشتباه (پرت)
normal_scores = np.random.normal(loc=15, scale=2, size=100)
outliers = np.array([85, 90])
all_scores = np.concatenate([normal_scores, outliers])

# رسم هیستوگرام
plt.figure(figsize=(8, 4))
plt.hist(all_scores, bins=60, color='skyblue', edgecolor='black')
plt.title("Histogram: Catching Outliers")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()
```

```python
# تولید داده‌های چوله (مثل توزیع ثروت یا حقوق)
salaries = np.random.exponential(scale=10, size=1000)

fig, ax = plt.subplots(figsize=(8, 4))

# رسم هیستوگرام و نمودار چگالی (KDE) برای نمایش بهتر چولگی
pd.Series(salaries).plot(kind='hist', bins=40, density=True, color='lightgreen', edgecolor='black', ax=ax, label='Histogram')
pd.Series(salaries).plot(kind='kde', color='blue', linewidth=2, ax=ax, label='KDE')

# رسم خط راهنما برای میانگین
plt.axvline(np.mean(salaries), color='red', linestyle='dashed', linewidth=2, label='Mean')

plt.legend()
plt.title("Histogram and KDE: Right Skewed Distribution")
plt.xlabel("Salary (in Millions)")
plt.show()
```

```python
# ترکیب قد ۵۰۰ زن و ۵۰۰ مرد
women_heights = np.random.normal(loc=160, scale=6, size=500)
men_heights = np.random.normal(loc=178, scale=7, size=500)
mixed_heights = np.concatenate([women_heights, men_heights])

# رسم هیستوگرام و نمودار چگالی پانداز برای نمایش بهتر قله‌ها
pd.Series(mixed_heights).plot(kind='hist', bins=40, density=True, color='thistle', edgecolor='black', figsize=(8, 4))
pd.Series(mixed_heights).plot(kind='kde', color='purple', linewidth=2)

plt.title("Histogram: Bimodal Distribution")
plt.xlabel("Height (cm)")
plt.show()
pd.Series(women_heights).plot(kind='hist', bins=40, density=True, color='thistle', edgecolor='black', figsize=(8, 4))
```
